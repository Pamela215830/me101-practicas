```python
import pandas as pd
import numpy as np
from scipy import stats
 
df = pd.read_csv("estudiantes_limpio.csv")
 
media = df["nota"].mean()
mediana = df["nota"].median()
std_muestral = df["nota"].std()                       # pandas: ddof=1
var_poblacional = np.var(df["nota"].to_numpy(), ddof=0)  # numpy: ddof=0
q1 = df["nota"].quantile(0.25)
q3 = df["nota"].quantile(0.75)
iqr = q3 - q1
 
print(f"Media: {media:.4f}")
print(f"Mediana: {mediana:.4f}")
print(f"Desv. estandar muestral: {std_muestral:.4f}")
print(f"Varianza poblacional: {var_poblacional:.4f}")
print(f"Q1: {q1:.4f}  Q3: {q3:.4f}  IQR: {iqr:.4f}")

```

```python
skew_pd = df["nota"].skew()
kurt_pd = df["nota"].kurt()
skew_sp = stats.skew(df["nota"])
kurt_sp = stats.kurtosis(df["nota"])
 
print(f"Skewness  pandas: {skew_pd:.4f} | scipy: {skew_sp:.4f}")
print(f"Kurtosis  pandas: {kurt_pd:.4f} | scipy: {kurt_sp:.4f}")
```

```python
def resumen_estadistico(serie, decimales=4):
    serie = serie.dropna()
    media = serie.mean()
    desv = serie.std()
    return {
        "n": int(serie.count()),
        "media": round(media, decimales),
        "mediana": round(serie.median(), decimales),
        "desv_std": round(desv, decimales),
        "cv_pct": round(desv / media * 100, decimales),
    }
 
print(resumen_estadistico(df["nota"]))
```

```python
def clasificar_dispersion(cv_pct):
    if cv_pct < 15:
        return "Baja"
    elif 15 <= cv_pct < 30:
        return "Moderada"
    else:
        return "Alta"


columnas = ["nota", "asistencia_pct"]

for columna in columnas:
    resumen = resumen_estadistico(df[columna])
    nivel = clasificar_dispersion(resumen["cv_pct"])
    print(
        f"Columna '{columna}': CV = {resumen['cv_pct']}% -> Dispersión {nivel}"
    )
```
