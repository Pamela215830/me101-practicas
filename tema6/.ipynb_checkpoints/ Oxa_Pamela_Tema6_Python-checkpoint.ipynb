```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Construir el degradado de 5x5
# np.linspace genera 5 valores equidistantes de 0 a 255.
# np.tile los repite en 5 filas.
fila = np.linspace(0, 255, 5)
gris = np.tile(fila, (5, 1))

# Impresión de propiedades
print("Shape:", gris.shape)
print("Dtype:", gris.dtype)
print("Mean:", gris.mean())

# 2. Visualizar y guardar la imagen
plt.imshow(gris, cmap="gray")
plt.colorbar()
plt.title("Degradado en Escala de Grises (5x5)")
plt.savefig("gris_tema6.png")
plt.show()

# 3. Recorte de la mitad derecha (columnas 3 a 4)
recorte = gris[:, 3:]
print("\nRecorte de la mitad derecha:")
print(recorte)
```

```python
import numpy as np

# 1. Construir la imagen RGB de dos mitades (4x4x3)
rgb = np.zeros((4, 4, 3), dtype=np.uint8)
rgb[:, 0:2] = [255, 0, 0]   # Mitad izquierda: Rojo (R=255, G=0, B=0)
rgb[:, 2:4] = [0, 0, 255]   # Mitad derecha: Azul (R=0, G=0, B=255)

print("Shape:", rgb.shape)
print("Pixel [0,0] (Rojo):", rgb[0, 0])
print("Pixel [0,3] (Azul):", rgb[0, 3])

# 2. Extraer cada canal
canal_R = rgb[:, :, 0]
canal_G = rgb[:, :, 1]
canal_B = rgb[:, :, 2]

print("\nCanal R:\n", canal_R)
print("Canal G:\n", canal_G)
print("Canal B:\n", canal_B)

# 3. Conversión a escala de grises
# Convertimos a float para evitar desbordamiento en operaciones algebraicas
R = rgb[:, :, 0].astype(float)
G = rgb[:, :, 1].astype(float)
B = rgb[:, :, 2].astype(float)

gris_promedio = (R + G + B) / 3.0
gris_luminosidad = 0.299 * R + 0.587 * G + 0.114 * B

print("\nGris Promedio:\n", gris_promedio)
print("\nGris Luminosidad:\n", gris_luminosidad)

# 4. Binarización con umbral propio
umbral = 50.0  # Elegido para separar la luminosidad del azul (29.07) de la del rojo (76.245)
binaria = gris_luminosidad > umbral

print("\nImagen Binarizada (Umbral = 50.0):\n", binaria)
```
