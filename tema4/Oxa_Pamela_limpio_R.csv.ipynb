```python
import pandas as pd
import numpy as np

datos = {
    "nombre":      ["Ana", "Luis", "Marco", "Rosa", "Iván", "Elena", "Diego", "Sofía"],
    "curso":       ["ME101", "ME101", "ME102", "ME101", "ME102", "ME102", "ME101",
"ME102"],
    "nota":        [16, 9, 18, np.nan, 14, 11, 8, np.nan],
    "asistencia_pct": [85, 60, 90, 75, np.nan, 95, 55, 80],
}
df_original = pd.DataFrame(datos)
df_original.to_csv("estudiantes.csv", index=False)
print("Archivo estudiantes.csv creado con", len(df_original), "filas")
```

```python
df = pd.read_csv("estudiantes.csv")
print(df.shape)
print(df.info())
df.head()
```

```python
print(df.isna().sum())
```

```python
df_reducido = df[["nombre", "nota", "curso"]]
aprobados = df[df["nota"] >= 10.5].copy()
aprobados["estado"] = "Aprobado"
print(aprobados[["nombre", "nota", "estado"]])
```

```python
df_limpio = df.dropna(subset=["nota"]).copy()
media_asistencia = df_limpio["asistencia_pct"].mean()
df_limpio["asistencia_pct"] = df_limpio["asistencia_pct"].fillna(media_asistencia)
print(df_limpio.isna().sum()) # debe ser 0 en todas las columnas
print(df_limpio.shape) # (6, 4): se eliminaron las 2 filas sin nota
```

```python
promedio_por_curso = df_limpio.groupby("curso")["nota"].mean()
print(promedio_por_curso)
```

```python
df_limpio.to_csv("estudiantes_limpio.csv", index=False)
```
