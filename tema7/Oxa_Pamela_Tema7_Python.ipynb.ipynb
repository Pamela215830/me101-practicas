```python
# Importación de librerías y creación de la imagen sintética
from PIL import Image, ImageDraw

# Crear imagen de 120x120 con fondo azul marino (RGB: 20, 30, 70)
img = Image.new("RGB", (120, 120), color=(20, 30, 70))
draw = ImageDraw.Draw(img)

# Figura 1: Círculo amarillo
draw.ellipse([20, 20, 60, 60], fill=(255, 215, 0))

# Figura 2: Rectángulo rojo
draw.rectangle([70, 60, 105, 100], fill=(220, 20, 60))

print("Tamaño original:", img.size, "| Modo:", img.mode)
img.save("mi_imagen.png")
import os
img.save("mi_imagen.png")
os.startfile("mi_imagen.png") 
```

```python
# Celda 2: Reabrir y verificar píxeles clave
img_reabierta = Image.open("mi_imagen.png")
print("Formato de la imagen:", img_reabierta.format)

# Píxeles clave anotados para uso posterior:
pix_fondo = img_reabierta.getpixel((5, 5))        # Fondo azul marino
pix_fig1 = img_reabierta.getpixel((40, 40))      # Círculo amarillo
pix_fig2 = img_reabierta.getpixel((85, 80))      # Rectángulo rojo

print(f"Píxel de fondo (5, 5): {pix_fondo}")
print(f"Píxel de Figura 1 (40, 40): {pix_fig1}")
print(f"Píxel de Figura 2 (85, 80): {pix_fig2}")
import os
os.startfile("mi_imagen.png") 
```

```python
#Realce de brillo con ImageEnhance
from PIL import ImageEnhance, ImageFilter

# Brillo menor a 1.0 (Oscurecer)
enhancer_dark = ImageEnhance.Brightness(img_reabierta)
img_dark = enhancer_dark.enhance(0.5)

# Brillo mayor a 2.0 (Aclarar / Saturar)
enhancer_bright = ImageEnhance.Brightness(img_reabierta)
img_bright = enhancer_bright.enhance(2.5)

# Píxel de la figura 1 antes y después del realce alto
pix_fig1_antes = img_reabierta.getpixel((40, 40))
pix_fig1_despues = img_bright.getpixel((40, 40))

print("Píxel Círculo (40,40) ANTES del realce (x2.5):", pix_fig1_antes)
print("Píxel Círculo (40,40) DESPUÉS del realce (x2.5):", pix_fig1_despues)
import os
os.startfile("mi_imagen.png")
```

```python
#Aplicación de Filtros y guardado
img_blur = img_reabierta.filter(ImageFilter.GaussianBlur(radius=3))
img_edges = img_reabierta.filter(ImageFilter.FIND_EDGES)

img_blur.save("mi_imagen_blur.png")
img_edges.save("mi_imagen_bordes.png")
print("Filtros aplicados y guardados exitosamente como PNG.")
import os
os.startfile("mi_imagen.png")
```
